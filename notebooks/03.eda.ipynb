# 03 - Exploratory Data Analysis (EDA)

## Objective

The objective of this notebook is to perform Exploratory Data Analysis
on the cleaned IBM HR Analytics Employee Attrition dataset.

EDA helps us understand patterns, relationships, distributions, and
trends within the dataset.

The main focus is to identify employee characteristics and workplace
factors associated with employee attrition.

### Analysis includes:
1. Univariate Analysis
2. Bivariate Analysis
3. Multivariate Analysis
4. Correlation Analysis
5. Business Insights and Findings

## 1. Importing Required Libraries

Import the required libraries for data manipulation, statistical
analysis, and data visualization.

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

sns.set_style('whitegrid')
```

## 2. Loading Cleaned Dataset

Load the cleaned dataset generated in the previous notebook.

```python
df = pd.read_csv('cleaned_hr_data.csv')

df.head()
```

## 3. Basic Data Understanding

Before performing detailed analysis, we examine the dataset's
statistical properties and identify the different types of features.

```python
df.shape
```

```python
df.describe()
```

## 4. Univariate Analysis

### 4.1 Employee Attrition Distribution

Analyze the distribution of employees who left the organization
and employees who stayed.

This helps us understand the overall attrition rate.

```python
attrition_count = df['Attrition'].value_counts()

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='Attrition')

plt.title('Employee Attrition Distribution')
plt.xlabel('Attrition')
plt.ylabel('Number of Employees')
plt.show()
```

```python
attrition_percentage = df['Attrition'].value_counts(normalize=True) * 100

print(attrition_percentage)
```

### 4.2 Age Distribution

Analyze the age distribution of employees to understand the
demographic composition of the workforce.

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='Age', bins=20, kde=True)

plt.title('Employee Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

### 4.3 Monthly Income Distribution

Analyze employee monthly income to understand the salary distribution
within the organization.

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='MonthlyIncome', bins=30, kde=True)

plt.title('Monthly Income Distribution')
plt.xlabel('Monthly Income')
plt.ylabel('Frequency')
plt.show()
```

### 4.4 Job Satisfaction Distribution

Analyze the distribution of employee job satisfaction levels.

Job satisfaction is measured on a scale from 1 to 4.

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='JobSatisfaction')

plt.title('Job Satisfaction Distribution')
plt.xlabel('Job Satisfaction Level')
plt.ylabel('Number of Employees')
plt.show()
```

### 4.5 Overtime Distribution

Analyze the number of employees who work overtime and those who do not.

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='OverTime')

plt.title('Employee Overtime Distribution')
plt.xlabel('Overtime')
plt.ylabel('Number of Employees')
plt.show()
```

## 5. Bivariate Analysis

### 5.1 Relationship Between Age and Attrition

Analyze how employee age is associated with attrition.

Compare the age distributions of employees who left and those who stayed.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='Attrition', y='Age')

plt.title('Age vs Employee Attrition')
plt.xlabel('Attrition')
plt.ylabel('Age')
plt.show()
```

### 5.2 Relationship Between Monthly Income and Attrition

Analyze whether monthly income differs between employees who left
and employees who stayed.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='Attrition', y='MonthlyIncome')

plt.title('Monthly Income vs Attrition')
plt.xlabel('Attrition')
plt.ylabel('Monthly Income')
plt.show()
```

### 5.3 Relationship Between Overtime and Attrition

Analyze whether employees working overtime have a different
attrition rate compared with employees who do not work overtime.

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='OverTime', hue='Attrition')

plt.title('Overtime vs Employee Attrition')
plt.xlabel('Overtime')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')
plt.show()
```

### 5.4 Relationship Between Department and Attrition

Analyze attrition across different departments to identify
departments with relatively higher employee attrition.

```python
plt.figure(figsize=(9, 5))

sns.countplot(data=df, x='Department', hue='Attrition')

plt.title('Department vs Employee Attrition')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=15)
plt.show()
```

### 5.5 Relationship Between Job Role and Attrition

Analyze employee attrition across different job roles to identify
roles with relatively higher attrition rates.

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    y='JobRole',
    hue='Attrition'
)

plt.title('Job Role vs Employee Attrition')
plt.xlabel('Number of Employees')
plt.ylabel('Job Role')
plt.show()
```

```python
jobrole_attrition = pd.crosstab(
    df['JobRole'],
    df['Attrition'],
    normalize='index'
) * 100

jobrole_attrition.sort_values('Yes', ascending=False)
```

### 5.6 Relationship Between Marital Status and Attrition

Analyze the relationship between employee marital status
and attrition.

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='MaritalStatus', hue='Attrition')

plt.title('Marital Status vs Employee Attrition')
plt.xlabel('Marital Status')
plt.ylabel('Number of Employees')
plt.show()
```

```python
marital_attrition = pd.crosstab(
    df['MaritalStatus'],
    df['Attrition'],
    normalize='index'
) * 100

marital_attrition
```

## 6. Multivariate Analysis

### 6.1 Job Satisfaction, Overtime and Attrition

Analyze how job satisfaction and overtime are jointly associated
with employee attrition.

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x='JobSatisfaction',
    y='MonthlyIncome',
    hue='Attrition',
    errorbar=None
)

plt.title('Job Satisfaction, Monthly Income and Attrition')
plt.xlabel('Job Satisfaction')
plt.ylabel('Average Monthly Income')
plt.show()
```

```python
sns.catplot(
    data=df,
    x='JobSatisfaction',
    hue='Attrition',
    col='OverTime',
    kind='count',
    height=4,
    aspect=1
)

plt.show()
```

### 6.2 Department and Job Satisfaction

Analyze average job satisfaction across departments and
compare the satisfaction levels of employees who left and stayed.

```python
department_satisfaction = df.pivot_table(
    index='Department',
    columns='Attrition',
    values='JobSatisfaction',
    aggfunc='mean'
)

department_satisfaction
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    department_satisfaction,
    annot=True,
    cmap='YlGnBu',
    fmt='.2f'
)

plt.title('Department vs Average Job Satisfaction')
plt.show()
```

### 6.3 Years at Company and Employee Attrition

Analyze the relationship between employee tenure and attrition.

This helps identify whether employees with shorter or longer
tenure have different attrition patterns.

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x='Attrition',
    y='YearsAtCompany'
)

plt.title('Years at Company vs Attrition')
plt.xlabel('Attrition')
plt.ylabel('Years at Company')
plt.show()
```

```python
df['TenureGroup'] = pd.cut(
    df['YearsAtCompany'],
    bins=[-1, 2, 5, 10, 20, 40],
    labels=['0-2 Years', '3-5 Years', '6-10 Years',
            '11-20 Years', '20+ Years']
)

tenure_attrition = pd.crosstab(
    df['TenureGroup'],
    df['Attrition'],
    normalize='index'
) * 100

tenure_attrition
```

## 7. Correlation Analysis

Correlation analysis helps identify the strength and direction of
linear relationships between numerical variables.

We use a correlation matrix and heatmap to visualize these relationships.

```python
numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(18, 12))

sns.heatmap(
    correlation_matrix,
    cmap='coolwarm',
    annot=False,
    linewidths=0.5
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
correlation_columns = numeric_df.drop(
    columns=['EmployeeNumber'],
    errors='ignore'
)

plt.figure(figsize=(18, 12))

sns.heatmap(
    correlation_columns.corr(),
    cmap='coolwarm',
    annot=False
)

plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

```python
df['Attrition_Encoded'] = df['Attrition'].map({
    'No': 0,
    'Yes': 1
})

attrition_correlation = (
    df.select_dtypes(include=np.number)
      .corr()['Attrition_Encoded']
      .sort_values(ascending=False)
)

attrition_correlation
```
